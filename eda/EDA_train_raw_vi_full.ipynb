```python
import os
import librosa
import pandas as pd
from tqdm import tqdm

def get_audio_metadata(root_dir, output_csv):
    data = []
    
    # Duyệt qua tất cả các tệp trong folder mẹ và folder con
    for subdir, _, files in os.walk(root_dir):
        for file in tqdm(files, desc=f"Processing {os.path.basename(subdir)}"):
            if file.endswith('.wav'):
                file_path = os.path.join(subdir, file)
                try:
                    # Lấy duration mà không cần load toàn bộ data vào RAM (giúp chạy nhanh hơn)
                    duration = librosa.get_duration(path=file_path)
                    
                    data.append({
                        'file_name': file,
                        'duration': duration
                    })
                except Exception as e:
                    print(f"Lỗi khi đọc file {file}: {e}")

    # Tạo DataFrame và xuất ra CSV
    df = pd.DataFrame(data)
    df.to_csv(output_csv, index=False)
    print(f"\nĐã lưu metadata vào: {output_csv}")

# Cấu hình đường dẫn
folder_me = r'D:\Speech_Verification\audio\train_vi_full'
file_output = r'D:\Speech_Verification\audio\train_vi_full_metadata.csv'

get_audio_metadata(folder_me, file_output)
```
