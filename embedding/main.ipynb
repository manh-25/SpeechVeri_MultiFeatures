### Trích xuất embedding giữ chiều thời gian

```python
from resume_embeeding import run_extraction_raw_and_subset, run_extraction_resume
import os
import glob
import torch

MODEL = "wav2vec2"

TRAIN_RAW_DIR = r"D:/Speech_Verification/audio/train_raw"

TRAIN_RAW_SAVE_DIR = f"F:/train_raw/{MODEL}_shards"

run_extraction_resume(
    model_key=MODEL,
    folder_path=TRAIN_RAW_DIR,
    save_dir=TRAIN_RAW_SAVE_DIR,
    batch_size=16,
    shard_size=200,
    time_cap=350,
    train_crop_mode="random",  # train: random crop khi T > 350
)

# Kiểm tra nhanh shard train_raw
raw_shard = sorted(glob.glob(os.path.join(TRAIN_RAW_SAVE_DIR, "*.pt")))[0]
raw_data = torch.load(raw_shard, map_location="cpu")
print("RAW:", raw_shard)
print("embeddings:", tuple(raw_data["embeddings"].shape))
print("lengths:", tuple(raw_data["lengths"].shape), "max_len:", int(raw_data["lengths"].max()))
print("nan:", torch.isnan(raw_data["embeddings"]).sum().item(), "inf:", torch.isinf(raw_data["embeddings"]).sum().item())
```

```python
from resume_embeeding import run_extraction_resume
import os
import glob
import torch

DATA_DIR = r"D:\my_project\SLP301-data\train_vi_5s"
MODELS = ["wav2vec2"]

for m in MODELS:
    save_dir = f"D:/my_project/SLP301-data/5s_embeddings/{m}_shards"

    run_extraction_resume(
        model_key=m,
        folder_path=DATA_DIR,
        save_dir=save_dir,
        batch_size=16,
        shard_size=5000,
        time_cap=350,
        train_crop_mode="random",  # train
    )

    shard_files = sorted(glob.glob(os.path.join(save_dir, "*.pt")))
    if not shard_files:
        raise FileNotFoundError(f"Không tìm thấy shard trong: {save_dir}")

    shard_file = shard_files[0]
    data = torch.load(shard_file, map_location="cpu")

    nan_count = torch.isnan(data["embeddings"]).sum().item()
    inf_count = torch.isinf(data["embeddings"]).sum().item()

    print(f"File: {shard_file}")
    print(f"Embeddings shape: {tuple(data['embeddings'].shape)}")   # (N, L, T_cap, D)
    print(f"Lengths shape: {tuple(data['lengths'].shape)}")
    print(f"Max length: {int(data['lengths'].max())}")
    print(f"Số lượng giá trị NaN: {nan_count}")
    print(f"Số lượng giá trị Inf: {inf_count}")
```
